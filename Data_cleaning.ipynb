```python
data['Delivery_person_ID'] = data['Delivery_person_ID'].str.upper()
```

```python
data['Delivery_person_Age'].unique()
```

```python
sns.kdeplot( data['Delivery_person_Age'] , fill =True )
```

```python
sns.boxplot(data=data, x='Delivery_person_Age')
```

```python
data['Delivery_person_Age'] = data['Delivery_person_Age'].fillna(data['Delivery_person_Age'].median())
```

```python
data['Delivery_person_Age'] = data['Delivery_person_Age'].astype(int)
```

```python
sns.kdeplot( data['Delivery_person_Ratings'] , fill = True )
```

```python
data['Delivery_person_Ratings'] = data['Delivery_person_Ratings'].fillna(data['Delivery_person_Ratings'].median())
```

```python
data['Order_Date'].unique()
```

```python
data['Order_Date'] = pd.to_datetime( data['Order_Date'], dayfirst = True )
```

```python
data['Order_Date'].unique()
```

```python
data['Vehicle_condition'].dtype
```

```python
data['Vehicle_condition'].unique()
```

```python
category_cols = ['Weather_conditions' ,  'Road_traffic_density' , 'Festival' ,  'City' ]
for col in category_cols:
    data[col] = data[col].fillna( data[col].mode()[0] )
```

```python
sns.kdeplot( data['multiple_deliveries'] , fill = True )
```

```python
sns.boxplot( data = data , x = 'multiple_deliveries' )
```

```python
data['multiple_deliveries'] = data['multiple_deliveries'].fillna(data['multiple_deliveries'].mode()[0])
```

```python
data['Time_Orderd'].unique()
```

```python
data = data.dropna(subset=['Time_Orderd'])
```

```python
((data.isnull().sum())/(len(data)))*100
```

```python
time_cols = ['Time_Orderd' , 'Time_Order_picked' , 'Time_taken (min)']
for col in time_cols:
    data[col] = pd.to_datetime( data[col] , format = "%H:%M:%S" )
```

```python
data['multiple_deliveries'] = data['multiple_deliveries'].astype(int)
```
